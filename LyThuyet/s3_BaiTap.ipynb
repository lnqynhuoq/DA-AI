### **2. Thống kê mô tả + số giá trị unique**

`describe()` tóm tắt cột số thành vài con số then chốt — không phân tích hết dataset.

`describe()` summarizes numeric columns into a few key numbers.

| Nhóm / Group | Chỉ số / Metrics |
|---|---|
| 1. Đầy đủ / Completeness | `count` |
| 2. Tâm / Center | `mean`, `median` (50%), `mode` |
| 3. Biên / Range | `min`, `max` |
| 4. Phân tán / Spread | `std`, `var`, `skew`, `kurt` |

```python
# TODO: gọi df.describe()
# df.describe(include="all")
df.describe(exclude="datetime") # loai datetime
df.describe(percentiles=[.05, .1, .15]) # liet ke theo % muon thay
```
